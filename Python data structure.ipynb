```python
# 20] Write a code to shuffle a given list randomly without using any built-in shuffle functions.
# Answer
def custom_shuffle(lst):
    n = len(lst)
    for i in range(n - 1, 0, -1):
        j = (i * 7 + 3) % (i + 1)
        lst[i], lst[j] = lst[j], lst[i]

s = [1, 2, 4, 5, 8]
custom_shuffle(s)
print(s)
```

```python
#31] Write a code that take a list of words as input  and returns a dictionary where the kays are unique words and the values
# are the frequencies of those words in te input list.
# Answer
def word_frequency(word_list):
    word_freq = {}
    for word in word_list:
        if word in word_freq:
            word_freq[word] += 1
        else:
            word_freq[word] = 1

    return word_freq
input_list = input("Enter a list of words separated by spaces: ").split()
result = word_frequency(input_list)
print(result)




```

```python
#32] Wrte a code that takes two dictionaries as input and merges theminto a single dictionary.If there arecommon keys,the values
# should be added together.
#Answer
def merge_dictionaries(dict1, dict2):
    merged = dict1.copy()  # Create a copy of the first dictionary
    for key, value in dict2.items():
        if key in merged:
            merged[key] += value  # Add values for common keys
        else:
            merged[key] = value  # Add new key-value pairs
    return merged

dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'b': 3, 'c': 4, 'd': 5}

result = merge_dictionaries(dict1, dict2)
print(result)


```

```python
# 33] Write a code that takes two dictionaries as input and manages them into a single dictionary.If there are common keys,the values
# should be added together
#Answer
def merge_dictionaries(dict1, dict2):
    # Creating a new dictionary to hold the merged results
    merged_dict = dict1.copy()  # Starting  with the first dictionary

    # Iterating over the second dictionary
    for key, value in dict2.items():
        if key in merged_dict:
            # If the key exists, adding the values together
            merged_dict[key] += value
        else:
            # If the key does not exist, adding the new key-value pair
            merged_dict[key] = value

    return merged_dict

dict1_input = input("Enter the first dictionary (e.g., a=1, b=2): ")
dict1 = dict(item.split('=') for item in dict1_input.split(','))
dict1 = {k: int(v) for k, v in dict1.items()}  # Converting values to integers

# Taking input for the second dictionary
dict2_input = input("Enter the second dictionary (e.g., b=3, c=4): ")
dict2 = dict(item.split('=') for item in dict2_input.split(','))
dict2 = {k: int(v) for k, v in dict2.items()}  # Convert values to integers

# Merging the dictionaries
result = merge_dictionaries(dict1, dict2)

print("Merged dictionary:", result)
```

```python
# 34] Write a code that takes a dictionary as input and returns a sorted version of it based on the values.You can choose whether to
# sort in ascending or descending order.
#Answer
def sort_dictionary_by_value(input_dict, descending=False):
    # Sorting the dictionary by value
    sorted_dict = dict(sorted(input_dict.items(), key=lambda item: item[1], reverse=descending))
    return sorted_dict

# Taking input for the dictionary
dict_input = input("Enter a dictionary (e.g., a=1, b=3, c=2): ")
input_dict = dict(item.split('=') for item in dict_input.split(','))
input_dict = {k: int(v) for k, v in input_dict.items()}  # Converting values to integers

# Asking user for the sorting order
order = input("Enter 'asc' for ascending or 'desc' for descending order: ").strip().lower()
descending = True if order == 'desc' else False

# Sorting the dictionary
sorted_dict = sort_dictionary_by_value(input_dict, descending)

# Printing the sorted dictionary
print("Sorted dictionary:", sorted_dict)
```

```python
# 35] Write a code that inverts a dictionary. swapping keys and values.Ensure that the inverted dictionary correctly handles cases
# where multiple keys have the same value by storing the keys as a list in the inverted dictionary.
# Answer
def invert_dictionary(input_dict):
    inverted_dict = {}

    # Iterating over the original dictionary
    for key, value in input_dict.items():
        # If the value is already a key in the inverted dictionary
        if value in inverted_dict:
            # Appending the current key to the list of keys for this value
            inverted_dict[value].append(key)
        else:
            # Creating a new entry with the value as the key and the current key as a list
            inverted_dict[value] = [key]

    return inverted_dict

# Taking input for the dictionary
dict_input = input("Enter a dictionary (e.g., a=1, b=2, c=1): ")
input_dict = dict(item.split('=') for item in dict_input.split(','))
input_dict = {k: int(v) for k, v in input_dict.items()}  # Converting values to integers

# Inverting the dictionary
inverted_dict = invert_dictionary(input_dict)

print("Inverted dictionary:", inverted_dict)
```
